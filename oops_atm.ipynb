```python
# ## ATM Machine


class SecureUser:

    def __init__(self, user_name, user_id, pin):
        self.user_name = user_name
        self.user_id = user_id
        self.__pin = pin
        self.amount = 0

    def validatePin(self, pin):
        if(self.__pin != pin):
         return 'Invalid Pin'
        return 'Your pin is correct'

class Atm(SecureUser):

  def checkBalance(self):
   return self.amount

  def depositAmount(self,amount):
    self.amount = amount
    return f"Dear {self.user_name}! Your amount {self.amount} is deposited successfully"

  def withdrawAmount(self,amount):
    if (self.amount < amount):
      return 'No Enough amount to withdaraw'
    else:
      self.amount -= amount
      return  f"Dear {self.user_name}! Available balance: {self.amount}"

# Dry rub

user = Atm("Vishwa", 101, 1234)

print(user.depositAmount(5000))
print(user.withdrawAmount(1000))
print(user.checkBalance())
print(user.validatePin(1234))

print("\n")
print("Example 2")

user1 = Atm("Janani", 102, 1001)

print(user1.withdrawAmount(1000))
print(user1.checkBalance())
print(user1.validatePin(1234))
```
