### 2.	Create a list of numbers and print only even numbers using a for loop.

```python
#2.	Create a list of numbers and print only even numbers using a for loop.

nums = [1,2,4,6,8,9,6,4,6,10]
nums = set(nums)
even_nums = []

for num in nums:
  if num % 2 == 0:
    even_nums.append(str(num))

if len(even_nums) > 0:
  print(f"Even numbers are {" ".join(even_nums)}")
else:
  print(f"No Even numbers are found")
```

### 3.	Write a program that handles division by zero using try-except.

```python
# 3.	Write a program that handles division by zero using try-except.
divident = int(input('Enter divident: '))
divisor  = int(input('Enter divisor: '))

try:
  quotient = divident / divisor
  print(f"quotient = {quotient}")
except ZeroDivisionError:
  print("Zero can't be divident")
```

###4.	Given a string, count how many vowels it has.

```python
# 4.	Given a string, count how many vowels it has.
text = 'PROITBRIDGE'
vowels = 'aeiou'
count = 0

for i in range(len(text)):
    if text[i].lower() in vowels:
        count += 1

print(f"Vowels present in {text} is {count}")
```

### 5. Reverse a list without using built-in functions.

```python
#5.	Reverse a list without using built-in functions.
nums = [1,2,3,4,5]

def reverse():
  reverse_num(0,len(nums)- 1,nums)
  return nums

def reverse_num(left,right,nums):
  if left > right:
    return nums
  nums[left], nums[right] = nums[right], nums[left]
  left +=1
  right -=1

reverse()
```

### 9.	Replace all missing values in a DataFrame with the column’s mean.

```python
# 9.	Replace all missing values in a DataFrame with the column’s mean.
import pandas as pd

data = {
   'Age': [25, None, 30, 35, None],
    'Salary': [50000, 60000, None, 80000, 70000]
}

df = pd.DataFrame(data)

#Replace missing values with column mean
df.fillna(df.mean(), inplace=True)

print(df)
```

### 10.	Filter rows in a DataFrame where the "Age" column is greater than 30.

```python
# 10.	Filter rows in a DataFrame where the "Age" column is greater than 30.
import pandas as pd

data = {
   'Age': [25, 40, 30, 35, 50],
    'Salary': [50000, 60000, None, 80000, 70000]
}

df = pd.DataFrame(data)
filtered_df = df[df['Age'] > 30]

print(filtered_df)
```
