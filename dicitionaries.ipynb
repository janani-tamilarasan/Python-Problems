## Built-in methods in Dictionaries

```
#create dictionaries
product = {
    "name": "Mechanical Keyboard",
    "price": 89.99,
    "in_stock": True,
    "sku": "KB-101"
}

print(product)
print(type(product))
```

```
#get(key, default)
print(f"get(key, default): {product.get('name')}")

#keys()
print(f"keys(): {product.keys()}")

#values()
print(f"values(): {product.values()}")

#items()
print(f"items(): {product.items()}")
```

```

product1 = {
    "name": "Mechanical Keyboard",
    "price": 89.99,
    "in_stock": True,
}


#setdefault(key, default)()
product.setdefault('qty', 2)
print(f"setdefault(key, default)(): {product}")

#update()
product.update(product1)
print(f"update(): {product}")

```
