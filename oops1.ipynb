## 1. Calculate the Student's Performance

Implement a class - Student - that has four properties and two methods. All these attributes (properties and methods) should be public. This problem can be broken down into three tasks.

1. Implement a constructor to initialize the values of four properties: name, phy, chem, and bio.

2.Implement a method totalObtained in the Student class that calculates total marks of a student.

3. Implement a method totalObtained in the Student class that calculates total marks of a student.





```python
# ## Calculate the Student's Performance
class Student:
  def __init__(self,name,phy,cem,bio):
    self.__name = name
    self.__phy = phy
    self.__cem = cem
    self.__bio = bio

  def totalObtained(self):
   return self.__phy + self.__cem + self.__bio

  def percentage(self):
    return self.totalObtained()/400 * 100

student = Student('Janani',60,70,50)
student.totalObtained()
student.percentage()
```

Implement this class according to the rules of encapsulation.

```python
#2. Implement the Complete Student Class
class Student:
  __name = None
  __rollNo = None

  def setName(self,name):
   self.__name = name

  def getName(self):
   return self.__name

  def setRollNo(self,rollNo):
   self.__rollNo = rollNo

  def getRollNo(self):
   return self.__rollNo


student = Student()
student.setName('Janani')
student.setRollNo('002')
print(student.getName(),student.getRollNo())

student1 = Student()
student1.setName('Vishwa')
student1.setRollNo('004')
print(student1.getName(),student1.getRollNo())

```
