## Lambda_Functions


### Lambda Function with map()


```python
# Exampe 1 -> Square of numbers
numbers = [1, 2, 3, 4, 5]

squares = list(map(lambda x: x ** 2, numbers))

print(squares)

# Example 2 - Convert uppecase to lowercase

fruits = ['APPLE','ORANGE', 'WATERMELON', "GRAPES"]

result = list(map(lambda fruit: fruit.lower() ,fruits))
print(result)
```

### Lambda Function with filter()

```python
## Example 1: Filter Even Numbers

numbers = [1, 2, 3, 4, 5, 6, 7, 8]

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print(even_numbers)


## Example 2 : Filter Names Starting with 'J'

names = ['Janani', 'john', 'Vishwa', 'Georje']

names_with_j = list(filter(lambda name: name.lower().startswith('j'),names))

print(names_with_j)
```

## Lambda Function with reduce()

```python
# Example 1: Find Sum of List Elements

numbers = [10, 20, 30, 40]

total = reduce(lambda x, y: x + y, numbers)

print(total)

# Example 2: Find Product of List Elements
numbers = [1, 2, 3, 4, 5]

product = reduce(lambda x, y: x * y, numbers)

print(product)
```
