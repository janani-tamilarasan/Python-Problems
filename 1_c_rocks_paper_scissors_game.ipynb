## Rocks, Paper, Scissors Game

Build a simple Rock-Paper-Scissors game where the player competes with the computer.This task will test your understanding of:

1. Input/output
2. Conditional logic
3. Random module
4. Loops
5. Exception handling

**Instructions**:

1.	The user should enter their choice: rock, paper, or scissors.

2.	The computer should randomly select one of these.

3.	Compare both and print who wins.

4.	Ask if the player wants to play again- if yes, repeat; if no, exit.

5.	Handle invalid inputs gracefully using try-except or condition checks.


```python
# Get Input with choices
# Validate choices
# Compare with random choice
# Repeat utill wins
# Ask for Replay if wins

import random
replay = True
while replay:
  choices = ('rocks', 'paper', 'scissors')
  random_choice = random.randint(1,3)
  try:
    user_choice = int(input("""
    Welome to Rocks,paper, siccors game
    Please provide you option to play:
    1. rocks
    2. paper
    3. scissors
    """))

    while int(user_choice) != int(random_choice):
      print(f"user_Choice {user_choice} random_choic {random_choice}")
      random_choice = random.randint(1,3)
      user_choice = input("Enter your option: ")
    print("YOU WON!!!!")

    # Replay option
    replay_choice = str(input('Do you want to play again- if yes, repeat; if no, exit.')).strip().lower()
    replay = replay_choice == 'yes'

  except ValueError:
     print("Invalid input. Please enter 1, 2, or 3.")
```
