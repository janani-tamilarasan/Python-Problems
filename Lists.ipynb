## 11 Built-in methods in List

```
cart = ['Pen','Fan','Book']

# Modification on List
# 1. Append(item)
cart.append('Bell')
print("append(item)): ", cart)

# Insert(index,item)
cart.insert(0,'IPhonr')
print("insert: ", cart)

# extend(iteratable)
batch_1 = [1, 2, 3]
batch_2 = [4, 5]
batch_1.extend(batch_2)
print("extend(iteratable): ",batch_1)
```

```
cart = ['Pen','Fan','Book']

# Removing item on List
# 1. remove(item)
cart.remove('Pen')
print("remove(item)): ", cart)

# pop()
cart.pop()
print("pop(): ", cart)

# clear()
cart.clear()
print("clear(): ", cart)
```

```
cart = ['Pen','Pen','Book']

# Accessing items on List
# 1. index(item)
cart.index('Pen')
print("index(item): ", cart.index('Pen'))

# count()
print("count(): ", cart.count('Pen'))
```

```
nums = [1,6,2.1,0.12,100,7]

# Order/copy items on List
# 1. sort()
nums.sort()
print("sort(): ", nums)

# reverse()
nums.reverse()
print("reverse(): ", nums)

# copy
alias_num = nums.copy()
nums[1] = 7
print('id: nums', nums)
print('id: alias_num', alias_num)
```
