## Data Visualization


### Load dataset

```python
import pandas as pd

import matplotlib.pyplot as plt

mtcars_data = pd.read_csv("https://raw.githubusercontent.com/janani-tamilarasan/Python-Problems/main/files/mtcars.csv")
```

### Histogram

### Bar Chart(Average MPG by Cylinders)

```python
## Bar Chart

avg_mpg = mtcars_data.groupby("cyl")["mpg"].mean()

plt.bar(avg_mpg.index, avg_mpg.values)

plt.title("Average MPG by Cylinders")
plt.xlabel("Cylinders")
plt.ylabel("Average MPG")

plt.show()
```

### Box Plot

```python
### Box Plot
import seaborn as sns

sns.boxplot(
    x="cyl",
    y="mpg",
    data=mtcars_data
)

plt.show()
```

### HEatmap

```python
plt.figure(figsize=(10,8))

sns.heatmap(
    mtcars_data.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm"
)

plt.show()
```

### Pair Plot


```python
sns.pairplot(
    mtcars_data[["mpg","hp","wt","disp"]]
)

plt.show()
```
