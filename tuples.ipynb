## Built-in methods in Tuples

```
#create a empty tuple
flowers = ()
print("empty tuple:", type(flowers))

# Tuple Packing: Python groups these integers into one tuple object
coordinates = 10, 20
print(f"Coordinates: {coordinates} (Type: {type(coordinates)})")

# Tuple Unpacking: Python distributes the tuple items into variables x and y
x, y = coordinates
print(f"x is {x}, y is {y}")
```

```
#count(value)
fruits = 'Apple','Oranges','Apple'
print(f"count: {fruits.count('Apple')}")

#index(value)
fruits = 'Apple','Oranges','Apple'
print(f"index: {fruits.index('Apple')}")
```
