##  Create 3 Dataframe using dictionaries
 - A DataFrame is a 2-dimensional table

```python
# Example 1 - Customer Data

import pandas as pd

data = {
     'Ids': [1,2,3],
     'Firstname':['Janani','Swamynathan','Vishwa'],
     'Lastname': ['Tamilarsan','Selvaraj','Swamynathan'],
     'Mobile': [00000000,None,None],
     'Email': ['janani@gmail.com','tst@gm.com','swany@gm.co'],
     'dob':  ['21/05/1990',None,'1/2/1984']
}

customer_data = pd.DataFrame(data)
customer_data
```

```python
# Example 2 - Products Data

import pandas as pd

data = {
     'ProductIds': [1,2,3],
     'ProdcutName':['Refirgerator','Washing Machine','Laptop'],
     'Qty': [2,1,0],
     'Dealer': ['Samsung','LG', 'HP'],
     'Stars':[None,None,3]
}

product_data = pd.DataFrame(data)
product_data
```
