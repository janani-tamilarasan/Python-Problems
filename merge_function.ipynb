## Merge Functions

```python
import pandas as pd

sales_data1 = pd.read_csv(
    "https://raw.githubusercontent.com/janani-tamilarasan/Python-Problems/main/files/sales_data1.csv"
)


sales_data2 = pd.read_csv(
    "https://raw.githubusercontent.com/janani-tamilarasan/Python-Problems/main/files/sales_data2.csv"
)


print("merged_data\n")
merged_data = pd.merge(sales_data1, sales_data2)

merged_data.info()


print("Concat\n")
concated_data = pd.concat([sales_data1, sales_data2],join= 'inner')

concated_data.info()
```
