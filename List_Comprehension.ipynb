## List Comprehension

new_list = [expression for item in iterable if condition]

### Example 1: Create a list of even numbers

```python
n = 30

even_numbers = [ num for num in range(1,n) if(num % 2 == 0)]

even_numbers
```

### Example 2: Create a list of squares

```python
numbers = [1, 2, 3, 4, 5]

squares = [num ** 2 for num in numbers]

print(squares)
```
