```python
import pandas as pd

data = pd.read_csv(
    "https://raw.githubusercontent.com/janani-tamilarasan/Python-Problems/main/files/sales_data.csv"
)

sales_data = pd.DataFrame(data)

# check missing values
sales_data.isnull().sum()


# remove missin values
sales_data = sales_data.dropna()

# check for duplicate
sales_data.duplicated(["Party", "Product"]).sum()
print(sales_data)

## Remove Dulicats
sales_data = sales_data.drop_duplicates()
print(sales_data)

## convert dat column
sales_data["Date"] = pd.to_datetime(sales_data["Date"])

## Rename column:
sales_data.columns = sales_data.columns.str.strip()

sales_data.rename(
    columns={
        "Qty": "Quantity"
    },
    inplace=True
)

## Invalid qtys
sales_data = sales_data[sales_data["Quantity"] > 0]

## Verify clean dta
print(sales_data.shape)
print(sales_data.isnull().sum())
print(sales_data.head())

# Highest sold prorducts

highest_sold = sales_data.groupby("Product")["Quantity"].sum().sort_values(ascending=False).head(1)
highest_sold

```
